## 2. Индексы и срезы

Срез `start:stop` берёт элементы с номерами от `start` до `stop` **не включая** `stop`. В матрице сначала пишут строки, потом столбцы: `M[строки, столбцы]`.

```python
print(M[0, 1])     # строка 0, столбец 1
print(M[1:3, 2:])  # строки 1–2, столбцы со 2-го до конца
print(M[:, -1])    # последний столбец
print(M[M > 6])    # все элементы больше 6
```

## 3. Векторизация: без циклов в 40 раз быстрее

Посчитаем MSE для миллиона предсказаний двумя способами.

```python
rng = np.random.default_rng(0)
y = rng.normal(size=1_000_000)
p = rng.normal(size=1_000_000)

start = time.perf_counter()
total = 0.0
for yi, pi in zip(y.tolist(), p.tolist()):
    total += (yi - pi) ** 2
loop_mse = total / len(y)
loop_time = time.perf_counter() - start

start = time.perf_counter()
vec_mse = np.mean((y - p) ** 2)
vec_time = time.perf_counter() - start
print(f"цикл: {loop_time:.3f} с, NumPy: {vec_time:.4f} с, быстрее в {loop_time / vec_time:.0f} раз")
print(np.isclose(loop_mse, vec_mse))
```
