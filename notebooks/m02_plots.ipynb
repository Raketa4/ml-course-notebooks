# Блокнот 2.6. Графики для разведочного анализа

Курс «Машинное обучение с нуля», урок 2.6. Сначала — квартет Анскомба, потом — пингвины.

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

anscombe = sns.load_dataset("anscombe")
penguins = sns.load_dataset("penguins")
```

## 1. Квартет Анскомба: сначала цифры

```python
anscombe.groupby("dataset").agg(mean_x=("x", "mean"), mean_y=("y", "mean"), std_y=("y", "std")).round(2)
```

```python
for name, part in anscombe.groupby("dataset"):
    print(name, round(np.corrcoef(part["x"], part["y"])[0, 1], 3))
```

Цифры почти одинаковые. А теперь посмотрим глазами:

```python
sns.relplot(data=anscombe, x="x", y="y", col="dataset", col_wrap=4, height=2.6)
plt.show()
```

## 2. matplotlib: фигура и оси

`fig, ax = plt.subplots()` создаёт фигуру и одну систему координат. Всё рисуем методами `ax`.

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(penguins["body_mass_g"].dropna(), bins=25, color="steelblue")
ax.set_xlabel("масса, г")
ax.set_ylabel("число пингвинов")
ax.set_title("Распределение массы")
plt.show()
```

## 3. seaborn: графики одной строкой

seaborn сам берёт столбцы таблицы по имени и раскрашивает по категории (`hue`).

```python
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue="species")
plt.show()
```

```python
sns.boxplot(data=penguins, x="species", y="body_mass_g")
plt.show()
```

```python
sns.pairplot(penguins, hue="species", height=1.8)
plt.show()
```
