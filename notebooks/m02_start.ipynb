## Задание 3. Первый график

```python
import matplotlib.pyplot as plt

distance = [1, 2, 3, 4]
fuel = [2.1, 3.9, 6.2, 7.8]
plt.scatter(distance, fuel, label="журнал полётов")
plt.plot([0, 4], [0, 2 * 4], color="orange", label="модель ŷ = 2·x")
plt.xlabel("расстояние, парсеков")
plt.ylabel("топливо, тонн")
plt.legend()
plt.show()
```
